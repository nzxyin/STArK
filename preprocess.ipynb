Generate Train dev test split files from sparc directory

```python
import os
import json
sparc_dirs = [
    '/data/user_data/xoy/LibriTTS_R/train-clean-360-sparc/emasrc/',
    '/data/user_data/xoy/LibriTTS_R/train-clean-100-sparc/emasrc/',
    '/data/user_data/xoy/LibriTTS_R/dev-clean-sparc/emasrc/',
    '/data/user_data/xoy/LibriTTS_R/test-clean-sparc/emasrc/',
]
output_paths = [
    '/data/user_data/xoy/LibriTTS_R/train-clean-360.json',
    '/data/user_data/xoy/LibriTTS_R/train-clean-100.json',
    '/data/user_data/xoy/LibriTTS_R/dev-clean.json',
    '/data/user_data/xoy/LibriTTS_R/test-clean.json',
]

# count the number of files in each directory
for sparc_dir in sparc_dirs:
    num_files = len(os.listdir(sparc_dir))
    print(f"{sparc_dir}: {num_files} files")

# create json files with just the filenames (without extensions)
for sparc_dir, output_path in zip(sparc_dirs, output_paths):
    filenames = [filename for filename in os.listdir(sparc_dir) if filename.endswith('.npy')]
    filenames_no_ext = [os.path.splitext(filename)[0] for filename in filenames]
    with open(output_path, 'w') as f:
        json.dump(filenames_no_ext, f)
    print(f"Created {output_path} with {len(filenames_no_ext)} entries")

# verify the contents of the created json files, print first 10 lines
for output_path in output_paths:
    with open(output_path, 'r') as f:
        data = json.load(f)
    print(f"Contents of {output_path} (first 10 lines):")
    for line in data[:10]:
        print(line)
```

```python
import random
import json
from pathlib import Path

# Config
metadata_path = "/data/user_data/xoy/LJSpeech-1.1/metadata.csv"
output_dir = "/data/user_data/xoy/LJSpeech-1.1/preprocessed/"
val_size = 150
test_size = 150
seed = 42

random.seed(seed)

# Read metadata and extract file IDs
file_ids = []
with open(metadata_path, "r", encoding="utf-8") as f:
    for line in f:
        if line.strip():
            file_id = line.split("|")[0]
            file_ids.append(file_id)

# Shuffle
random.shuffle(file_ids)

# Split
test_ids = file_ids[:test_size]
val_ids = file_ids[test_size:test_size + val_size]
train_ids = file_ids[test_size + val_size:]

# Create output directory
Path(output_dir).mkdir(parents=True, exist_ok=True)

# Save JSON files
def save_json(filename, data):
    with open(Path(output_dir) / filename, "w", encoding="utf-8") as f:
        json.dump(data, f, indent=2)

save_json("train.json", train_ids)
save_json("val.json", val_ids)
save_json("test.json", test_ids)

# Summary
print(f"Total: {len(file_ids)}")
print(f"Train: {len(train_ids)}")
print(f"Val: {len(val_ids)}")
print(f"Test: {len(test_ids)}")
```
